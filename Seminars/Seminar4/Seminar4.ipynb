## 1.1. Зачем нужно предсказывать структуры белков?  

* **Понимание функции белка**  
    Пространственная структура определяет, как белок выполняет свои биологические функции и взаимодействует с другими молекулами. Без знания структуры невозможно полностью понять молекулярные механизмы, лежащие в основе активности белка, его динамики и конформационных изменений.


    <img src="Images/struc_dyn_func.png" alt="drawing" style="width:200px;"/> 

* **Изучение взаимодействий с лигандами и другими белками**  
    3D-структура позволяет выявлять специфические участки связывания и регуляции, что необходимо для понимания механизмов действия белков, их комплексообразования и участия в сигнальных путях.

* **Разработка новых лекарственных препаратов**  
    Структурная информация активно используется для дизайна и оптимизации биологически активных соединений. она облегчает выявление активных и аллостерических сайтов для взаимодействия с потенциальными лигандами, оптимизацию молекул по эффективности и селективности.
  
* **Ферментная инженерия**  
    Предсказание структуры позволяет модифицировать ферменты для улучшения их катализаторных свойств, стабильности и специфичности, что важно для биотехнологии, промышленности и медицины.

## 1.3. Методы in silico предсказания структуры белков

<img src="Images/researchers-apply-quan-1.jpg" alt="drawing" style="width:900px;"/> 

Предсказание структуры белков in silico включает два больших класса методов: 
* template-based (основанные на шаблонах)
* physics-based (физически обоснованные)

### 1.3.1. Template-based methods

* **Homology modeling (моделирование по гомологии)**
    * Моделирование по гомологии строит 3D структуру целевого белка на основе известной структуры похожего (гомологичного) белка — шаблона. 
    * Метод опирается на то, что эволюционно близкие белки имеют сходную пространственную структуру, даже при значительном различии последовательностей аминокислот. 
    * Сначала ищут известную структуру — шаблон, затем проводят выравнивание последовательностей целевого белка и шаблона и строят модель целевого белка на основе структуры шаблона. 
    * Метод эффективен при высокой идентичности последовательностей (>30%)


* **Threading (Fold recognition)**
  * Метод пытается «нащупать» подходящую структурную основу, даже если последовательности схожи слабо или совсем не похожи. 
  * Threading сопоставляет последовательность целевого белка с базой структур и выбирает ту, которая лучше всего подходит по физико-химическим свойствам и структурным мотивам.

* **Deep learning methods**
  
  Настоящий прорыв произошёл с использованием нейросетей:

  * **AlphaFold2 (DeepMind)** — модель, которая на базе многомасштабного обучения предсказывает структуру с точностью, близкой к экспериментальным методам.
  * **RoseTTAFold** — разрабатываемый в лаборатории Дэвида Бейкера, три-трековая нейронная сеть, которая одновременно моделирует последовательности, пары остатков и трехмерное расположение.
  * Эти методы не требуют жесткой зависимости от близких шаблонов и могут эффективно предсказывать структуру для широкого класса белков.

### 1.3.2. Physics-based methods

* **Ab initio modeling** 
  * Предсказание структуры основано строго на физико-химических принципах — минимизации свободной энергии, универсальных силах взаимодействия и вычислительной симуляции свёртывания белка из последовательности.
  * Этот метод не нуждается в знакомых шаблонах и пытается вычислить наиболее энергетически выгодную конфигурацию. Подходит для маленьких белков из-за чрезвычайной вычислительной сложности.

* **Молекулярная динамика (MD)**
  
  * MD моделирует движение белковых атомов во времени под действием физических сил, позволяя изучать стабильность, гибкость и конформационные переходы. 
  * MD часто используется для доработки моделей и исследования динамических аспектов структуры, но само по себе, чаще всего, не является методом первичного предсказания структуры белка.

* **Квантовые алгоритмы**
  * Современные разработки исследуют гибридные классические-квантовые методы для точного моделирования белковых систем и свёртывания на уровне электронных взаимодействий, что может в будущем повысить качество предсказаний и охватить более сложные системы.

### 1.6.2. Этапы моделирования по гомологии

<img src="Images/image-1059.png" alt="drawing" style="width:900px;"/> 

1. **Идентификация шаблонов**: 
   * Идентификация одного или нескольких белков-шаблонов с известными экспериментальными структурами и значительным сходством последовательностей с целевым белком
   * Выбор шаблонов должен проводиться с осторожностью, чтобы гарантировать их структурное соответствие исследуемому белку
   
2. **Выравнивание последовательностей** : 
   * Выравнивание последовательностей необходимо для определения соответствующих аминокислотных позиций в целевом белке и белке(ах)-шаблоне(ах). 
   * Процесс выравнивания играет решающую роль в определении связи между последовательностями-шаблоном и целевой последовательностью.
   
3. **Подготовка структуры шаблона**: 
   * Подготовка структуры шаблона включает в себя подготовку структур шаблонов, которые будут использоваться в процессе моделирования. 
   * Этот процесс может включать в себя удаление посторонних компонентов из архитектуры шаблона, таких как молекулы воды или лиганды, и проверку соответствия структуры шаблона требуемому формату для моделирования
   
4. **Построение модели**: 
   Процесс построения модели включает в себя: 
   *  создание исходной трёхмерной модели целевого белка путём переноса структурной информации из шаблонного белка (белков) в целевой белок. Этот перенос основан на выравнивании позиций последовательностей. 
   *  корректировку структуры шаблона (или шаблонов) в соответствии с целевой последовательностью, а затем размещение соответствующих остатков в модели.

5. **Моделирование петель и размещение боковых цепей**: 
   * Процесс моделирования петель и размещения боковых цепей включает моделирование любых отсутствующих петель или гибких участков в исследуемой структуре белка. 
   * Этот этап подразумевает прогнозирование конформации участков петель, отсутствующих в шаблонной конфигурации. 
   * Кроме того, рекомендуется размещать боковые цепи аминокислот в модели таким образом, чтобы максимизировать локальные взаимодействия и энергетику.
   
6. **Уточнение модели**: 
   * Уточнение модели включает в себя процесс повышения качества и точности исходной модели. 
   * Оптимизация геометрии модели, повышение её стабильности и общего качества могут быть достигнуты с помощью различных методов, таких как минимизация энергии, моделирование молекулярной динамики и другие методы оптимизации.
   
7. **Оценка и валидация модели**: 
   * Процесс оценки и валидации модели включает в себя оценку качества и надёжности итоговой модели
   * Валидация модели может осуществляться с использованием различных методов и критериев, таких как оценка энергии модели, проверка правильности стереохимии и сопоставление модели с имеющимися экспериментальными данными. 
   * Процесс валидации служит для проверки биологической достоверности и надёжности модели для последующего анализа.

### 2.1.2. Выбор шаблонов

После завершения поиска шаблона качество шаблона оценивается двумя методами: 

1. **Глобальная оценка качества модели (GMQE)**
   * GMQE объединяет свойства, полученные из выравнивания целевой модели и шаблона, и структуру шаблона, и выражает ожидаемую точность или надёжность модели. 
   * GMQE принимает значения от 0 до 1, где 1 соответствует максимальной точности, а 0 — минимальной.

2. **Оценка качества четвертичной структуры (QSQE)**
   * Оценка QSQE также принимает значения от 0 до 1, однако она вычисляется только для шаблонов с наивысшим рейтингом, если существует вероятность построения олигомера. 
   * Значение выше 0,7 считается надёжным.



Кроме того, для каждого шаблона указаны: 

* **Экспериментальный метод**, с помощью которого была получена структура шаблона. 
  
  * Как правило, при моделировании по гомологии рекомендуется использовать рентгеновские кристаллические структуры с разрешением ниже 2,2Å в качестве шаблонов. 
  * Однако, часто приходится идти на компромисс между высокой идентичностью/сходством последовательностей и разрешением шаблона. 
  * В целом, структуры, полученные методом рентгеновской кристаллографии, предпочтительнее усреднённых структур ЯМР и структур, полученных с помощью электронной микроскопии, поскольку последняя определяет общую форму молекулы, а не положение отдельных атомов.
  
* **Метод поиска**, которым был найден шаблон (Blast/HHblits) 
  
* **Sequence similarity** между последовательностью и шаблоном рассчитывается с помощью матрицы замен BLOSUM62 (1 соответствует 100% сходству последовательностей, 0 - наоборот). При расчете сходства последовательностей пробелы не учитываются.

* **Лиганды** в структуре шаблона - может быть важно для моделирования структуры в различных состояниях связывания целевого белка с лигандами
  
* **Олигомерное состояние** - прогнозируется для каждого шаблона, но пользователь может изменить его вручную в разделе «target prediction». Если олигомерное состояние модели не полностью соответствует состоянию шаблона, то появится предупреждающий знак.
  
* **Выравнивание** последовательностей между исходной последовательностью и шаблоном

    * Консервативная характеристика интерфейса может быть весьма полезна для определения того, насколько хорошо интерфейсы шаблонов адаптируются к целевому семейству белков: близкие гомологи должны достигать самых низких значений консервативности интерфейса при максимально возможном пороговом значении идентичности.

### 2.1.2. Building models

1. Чтобы построить модель из выбранного шаблона, сначала переносятся идентичные координаты атомов, моделируются вставки и неконсервативные боковые цепи аминокислот. 
   * Этот шаг выполняется движком моделирования ProMod3, который основан на фреймворке вычислительной структурной биологии OpenStructure
   * ProMod3 извлекает структурную информацию из выровненной структуры шаблона в декартовом пространстве, и, если подходящие фрагменты не находятся, используется выборка Монте-Карло для выполнения поиска в конформационном пространстве. 
  
2. Новые конформации боковых цепей, которые не могут быть найдены в шаблоне, моделируются на основе библиотеки ротамеров, зависящих от основной цепи
   
3. На заключительном этапе небольшие структурные искажения или стерические столкновения разрешаются путем короткой минимизации энергии с использованием силового поля CHARMM27 .

Как и в при выборе шаблона, список построенных моделей отображается в разделе «Результаты модели»


* GMQE (Global Model Quality Estimation) 
  * ожидаемая оценка качества модели белка, учитывающая выравнивание целевой последовательности с шаблоном и покрытие модели. 
  * Значения варьируют от 0 до 1 (более высокие значения указывают на лучшее ожидаемое качество модели). 
  * GMQE полезен для выбора оптимальных шаблонов до построения модели и комбинирует характеристики выравнивания и структуры шаблона в предсказании качества модели.​
  *  value above 0.7 is considered reliable.


* QMEANDisCo
  * комплексная функция оценки качества модели белка, которая дает как глобальную (для всей структуры), так и локальную (для каждого остатка) абсолютные оценки качества. 
  * Она основана на классической оценке QMEAN, дополненной прогнозом качества для каждого остатка, основанным на сопоставлении парных расстояний между остатками с констрейнтами, извлеченными из гомологичных структур. 
  * Для объединения этих данных используется нейронная сеть, обученная на предсказании локальной меры качества lDDT в диапазоне от 0 до 1. 
  * Глобальный QMEANDisCo — среднее значение по всем остаткам, с оценкой ошибки для модели данного размера.​
  *  A good quality prediction is expected to have a QMEANDisCo above 0.70

Таким образом, GMQE служит предварительной оценкой качества на основе шаблона и выравнивания, а QMEANDisCo — более детализированной оценкой качества конкретной построенной модели с учетом информации о гомологических структурах и геометрических характеристиках модели.​
