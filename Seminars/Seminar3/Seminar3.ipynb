# 1. Последовательности в биологии

## Какие бывают последовательности?

1. **Нуклеотидные (ДНК/РНК)**: 
   * **Нуклеотид** — это строительный блок ДНК и РНК. Он состоит из трех частей: фосфатной группы, сахара (дезоксирибозы в ДНК и рибозы в РНК) и азотистого основания (аденин, тимин/урацил, гуанин или цитозин в ДНК/РНК). Нуклеотиды соединяются друг с другом, образуя цепочку, которая и составляет структуру ДНК или РНК.

<img src="Images/DNA1.png" alt="drawing" style="width:600px;"/> 



<img src="Images/DNA.png" alt="drawing" style="width:300px;"/> 


2. **Аминокислотные (белки)**
   * **Аминокислота** - строительный блок белка. Она состоит из аминогруппы, карбоксильной группы и бокового радикала. Аминокислоты соединяются между собой с помощью пептидной связи. 

<img src="Images/peptide-chain.png" alt="drawing" style="width:900px;"/> 


## Как связаны эти последовательности?

Через центральную догму молекулярной биологии **ДНК➔РНК➔белок**

<img src="Images/Central-Dogma-DNA-to-RNA-to-Protein.png" alt="drawing" style="width:500px;"/> 


Как связаны нуклеотиды и аминокислоты? 

* **Триплет** — комбинация из трех нуклеотидов, она кодирует одну аминокислоту. 
  
  64 возможные комбинации (4\*4\*4) соответствуют всего 20 аминокислотам, поэтому кодонная система имеет избыточность, и разные триплеты могут кодировать одну и ту же аминокислоту.

<img src="Images/Why genetic code is a triplet code logic diagram.jpg" alt="drawing" style="width:600px;"/> 


* **Ген** — это участки ДНК, несущие целостную информацию о строении одной молекулы РНК/белка.  

* **Геном** — полный набор генов организма.

# 3. Выравнивание последовательностей

**Выравнивание последовательностей** — биоинформатический метод, основанный на размещении двух или более последовательностей мономеров ДНК, РНК или белков друг под другом таким образом, чтобы легко увидеть сходные участки в этих последовательностях. 

Зачем сравнивать последовательности (аминокислотные и нуклеотидные):
  * Найти похожие последовательности в базе данных
  * Для поиска мутаций в популяции или наборе генов
  * Найти консервативные для разных организмов участки, ключевые для активности белка
  * Определить вариабельные участки, которые могут объяснять различия функций белков
  * Проследить изменения, обусловленные эволюцией, которые влияют на свойства белков

## Типы выравниваний


* **Попарное**: 
  
  * **Глобальное**- сравнивает две последовательности по всей длине. Оно лучше всего подходит для очень похожих последовательностей примерно одинаковой длины.
  
  * **Локальное** - позволяет найти наиболее схожие участки в двух последовательностях. Может быть получено более одного выравнивания. Этот метод лучше всего подходит для последовательностей с некоторой степенью сходства или разной длины. Наиболее подходит для поиска консервативных элементов.
  
* **Множественное** (MSA - multiple sequence alignment)

## BLAST

**BLAST (Basic Local Alignment Search Tool)** - самая известная эвристика, основанных на локальном выравнивании по алгоритму Смита-Ватермана (*существуют и другие, например, FastA*)

<img src="Images/Blast.png" alt="drawing" style="width:700px;"/> 

### Матрицы замен аминокислот

**Protein BLAST** подсчитывает баллы за совпадения и несовпадения аминокислот, используя предопределенные матрицы PAM/BLOSUM.

* **PAM (Point Accepted Mutations)**
  * Частоты замен аминокислот, полученные из выравниваний родственных белков (не менее чем 85 % сходства)
    * PAM1 соответствует изменению 1 аминокислоты на 100 аминокислот (1% замен).
    * PAM100 – 100 изменений на 100 аминокислот; $(PAM1)^100$.
    * PAM250 – 250 изменений на 100 аминокислот; $(PAM1)^250$.


* **BLOSUM (BLOcks SUbstitution Matrices)**
  * Анализировали консервативные области последовательностей родственных белков, полученные из базы данных BLOCKS.
  * Разделили белки на группы по мере схожести:
    * Например, BLOSUM62 - группы состояли из белков, для которых уровень идентичности последовательностей >62%
    * BLOSUM90 ~ 90%
  * Расчитали среднюю вероятность замен внутри каждой группы, преобразовали в веса матрицы 

Более подробно: https://bioinformaticshome.com/bioinformatics_tutorials/sequence_alignment/substitution_matrices.html

# 6. Филогенетические деревья

**Филогенетическое дерево** - это визуальное представление о гипотетическом ходе эволюции в виде графа.

## Строение филогенетического дерева: 



<img src="Images/PhyloTree.jpg" alt="drawing" style="width:400px;"/>

## Отношения в филогенетическом дереве: 

* **Гомологи** - это гены/белки, имеющие общего предка

  * **Ортологи** - это гомологи из разных организмов, разошедшиеся в процессе видообразования и обычно выполняющие одну и ту же функцию.
  * **Паралоги** - это гомологи из одного организма, возникшие из общего предка в результате дупликации генов. Как правило, имеют разные функции.

<img src="Images/homolog-ortolog-paralog-genler-8958.png" alt="drawing" style="width:600px;"/>

## Топология дерева: 



* **Дендрограмма** - отражает только топологию (взаимное расположение клад), длина ветвей не несет информации
* **Филограмма** - длина ветвей пропорциональна эволюционному расстоянию между узлами
* **Хронограмма** - длина ветвей соответствует эволюционному времени


<img src="Images/Phylotrees_types.png" alt="drawing" style="width:600px;"/>


## Виды деревьев


* **Укорененное дерево** — содержит выделенную вершину — корень. Каждый узел такого дерева соответствует  ближайшему общему предку (most recent common ancestor) нижележащих листьев дерева.
* **Неукороенное дерево** - нет корня: показывают родство организмов без указания происхождения.

### 1. Методы расстояний


Общий принцип заключается в попарном сравнении выровненных последовательностей и построении матрицы расстояний, которая затем используется для построения филогенетического дерева. 

Эти методы простые и быстрые, однако имеют недостаток - сводят данные о сходстве/различиях в последовательностях к одному числу (расстоянию), что приводит к потере информации. 

* UPGMA - невзвешенного попарного среднего
* **NJ - присоединения ближайшего соседа**
* MS - наименьших квадратов
* ME - минимальной эволюции

#### Расстояние между аминокислотными последовательностями

* **Нескорректированное расстояние (percent identity)**:
  * Простые расчеты, при которых подсчитывается количество позиций аминокислот, которые различаются в двух последовательностях
  * Расстояние часто выражается в процентах идентичности.

* **Методы матрицы замен**:
  * Используют матрицы оценок (например, PAM или BLOSUM), которые отражают вероятность замены одной аминокислоты другой на основе эволюционных факторов.
  * Такие подходы позволяют учитывать множественные мутации, происходящие в одной и той же позиции, и тот факт, что некоторые замены аминокислот более вероятны, чем другие.

> Более подробно: https://en.wikipedia.org/wiki/Substitution_model#Amino_acid_substitution_models

#### UPGMA (Unweighted Pair Group Method with Arithmetic Mean)

* Самый простой метод
* Необходимым условием использования метода является постоянная скорость эволюции исследуемых последовательностей (гипотеза **"молекулярных часов"**). При неравномерной скорости эволюции метод может приводить к ошибкам в топологии дерева.
* Генерирует укоренённые деревья 



**Алгоритм** (см. схему ниже):

>1. Находим два самых близких значения в матрице ($B$ и $F$)
>2. Объединяем соответствующие узлы в группу ($BF$)
>3. Пересчитываем расстояния до остальных узлов как среднее расстояние до первого и второго членов группы 
>$d(X, BF) = 0.5 \cdot (d(X, B) + d(X, F)) $
>4. Строим новую матрицу (-1 строка, -1 столбец)
>5. Если матрица все еще больше 1x1, см. п. 1

<img src="Images/UPGMA.png" alt="drawing" style="width:600px;"/>


#### NJ (Neighbor-Joining)

* Метод основывается на предположении, что наиболее вероятным будет дерево с наименьшим количеством эволюционных событий. 
* Генерирует неукорененные деревья

**Алгоритм**: 

>1. Начинаем со "звездной" конфигурации дерева
>2. По матрице расстояний находим ближайшую пару узлов (либо A-B, либо D-E)
>3. Пересчитываем новую матрицу (-1 строка, 1 столбец)
>4. Если матрица все еще больше 1x1, см. п. 2
