## 2.1. Аминокислоты

* Аминокислоты являются мономерами белков
* Каждая аминокислота состоит из: 
    * аминогруппы (-NH2)
    * карбоксильной группы (-COOH)
    * боковой цепи (R-группы) и водорода (H), присоединенных к центральному атомному углероду (Cα)
    
Rem: Валентности углерода, азота и кислорода равны 4, 3 и 2 соответственно.

## 2.2. Образование полипептидной цепи

* Пептидная связь образуется между аминогруппой одной аминокислоты и карбоксильной группой другой аминокислоты в результате реакции дегидратации, при которой выделяется молекула воды (H₂O), а между атомами C и N образуется ковалентная связь.  

<img src="Images/Peptide-Bond-01.png" alt="drawing" style="width:500px;"/> 

## 2.2. Торсионные углы в полипептидной цепи

- Торсионные углы (dihedral angles) – это углы, характеризующие вращение вокруг связей в белковой цепи. Основные углы включают:
  - **Угол фи (φ)**: вращение вокруг связи между N-атомом аминогруппы и Cα-атомом.  
  - **Угол пси (ψ)**: вращение вокруг связи между Cα-атомом и C-атомом карбоксильной группы.  
  - **Угол омега (ω)**: вращение вокруг пептидной связи (между C-концом и N-концом соседних аминокислот).  


  <img src="Images/Dihedral-angles_withw.png" alt="drawing" style="width:500px;"/> 

## 2.3. Диаграмма Рамачандрана

Карта Рамачандрана — это диаграмма визуализации двугранных углов аминокислот полипептидной цепи (ψ и φ). 
* точки соответствуют аминокислотным остаткам, по горизонтали откладывается угол φ, а по вертикали — ψ, точки 
* позволяет с высокой степенью точности определить вторичную структуру белка: с помощью карты можно установить, какие торсионные углы характерны для α-спиралей и для β-листов, для различных аминокислотных остатков

# 3. Практика в PyMOL

## 3.1. Модуль Builder

- Панель модуля Builder для настройки параметров

- Команда для показа атомов в виде сфер: `show spheres`.  
  - Изменить размер сфер: **Settings -> Edit all -> sphere -> sphere scale -> 0.25** или команда `set sphere_scale, 0.25`.  

- Направление синтеза белков: цепь белка растет от N-конца (аминогруппы) к C-концу (карбоксильной группе).  
  - Демонстрация в Builder: создание трёх аминокислот аланина (ALA -> ALA -> ALA).  

- Скрытие каркасной структуры и отображение атомов в виде сфер:  
  - `hide cartoon`, `show spheres`, `hide (hydro)`.

## 3.2. Модуль Wizard

- **Measurement -> Dihedrals** для расчета торсионных углов.  

- **Labels -> Atom name** - для отображения названий атомов.  

- Пример расчета углов для структуры:  
  - Угол φ: выбрать атомы C1, N1, Calpha2, C2.  
  - Угол ψ: выбрать атомы N1, Calpha1, C1, N2.
  - Угол ω: выбрать атомы Calpha1, C1, N2, Calpha2.  

- Команда `get_dihedral atom1, atom2, atom3, atom4` https://wiki.pymol.org/index.php/Get_Dihedral

## 3.3. Изменение двугранных углов
