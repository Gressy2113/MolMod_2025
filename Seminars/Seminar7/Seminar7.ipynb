# 1. Введение

* **Биологические мембраны** состоят из липидного бислоя толщиной ~5–10 нм, в который встроены белки.
  * Многие белки и липиды внешнего слоя дополнительно несут углеводные цепи; совокупность гликопротеинов, гликолипидов и связанных полисахаридов формирует **гликокаликс** клеточной поверхности. 
  * Соотношение липиды:белки:углеводы существенно зависит от типа клетки и конкретной мембраны, но чаще всего основную массу дают липиды и белки

* Липидный бислой имеет **жидкостно‑мозаичную структуру**: 
  * липиды и белки диффундируют в плоскости мембраны, тогда как «флип‑флоп» переходы липидов между листами бислоя более редки и требует участия специальных ферментов - транслоказ (исключение — холестерин, который перескакивает легче). 
  * Текучесть мембраны определяется составом липидов и температурой: ненасыщенные хвосты за счёт двойных связей нарушают упаковку и повышают подвижность


<img src="Images/Membrane.png" alt="drawing" style="width:800px;"/> 

*Строение биологической мембраны https://lifelib.info/biochemistry/leninger/*

* **Функции мембран**
  * *Ограждение и изоляция клеток и органелл* - создают физический и химический барьер, поддерживают баланс веществ между внутри- и внеклеточной средой
  * *Регуляция транспорта веществ* - контролируют поступление и выход ионов и молекул, обеспечивая гомеостаз и избирательный транспорт
  * *Восприятие внеклеточных сигналов* - содержат рецепторы, которые распознают сигналы и запускают внутриклеточные каскады
  * *Ферментативный катализ* - на мембранах локализованы ферменты, в том числе участвующие в энергетическом метаболизме и биосинтезе липидов
  * *Взаимодействие с другими клетками* - обеспечивают адгезию, распознавание и слияние клеток, связь с внеклеточным матриксом
  * *Закрепление элементов цитоскелета* - служат опорой для элементов цитоскелета, что важно для формы клетки и транспорта

### 1.1. Состав мембраны

* **Мембранные липиды**: 
  * **Фосфолипиды** (фосфатидилхолин, фосфатидилэтаноламин, фосфатидилсерин, фосфатидилинозитол, сфингомиелин), **холестерин** и **гликолипиды**; все они амфифильны и в воде самопроизвольно образуют бислои за счёт гидрофобного эффекта и ван‑дер‑ваальсовых взаимодействий. 
  * Фосфолипиды и гликолипиды имеют полярные гидрофильные головки и две гидрофобные углеводородные цепи, а углеводные фрагменты гликолипидов ориентированы в водную фазу.

## 2.3. Расчет МД на кластере

### Пайплайн уравновешивания

* **Цель уравновешивания**: постепенно снять ограничения на движение атомов, чтобы избежать артефактов и "поломки" структуры после сборки.

* **Типы ограничений**: 

  * **BB**: позиционные ограничения на остов белка (backbone).
  * **SC**: позиционные ограничения на боковые цепи белка (sidechains).
  * **wforce**: позиционные ограничения на воду (обычно по оси Z).
  * **tforce**: позиционные ограничения на хвосты липидов (tails).
  * **mforce**: позиционные ограничения на головные группы липидов ("mid"/"headgroup").
  * **ion**: позиционные ограничения на ионы (Na+, Cl^-) — только на старте.

* Позиционные ограничения прописывают в файлах топологий `.itp`, например, в файле `PROA.itp`: 
    ```
    #ifdef POSRES
    [ position_restraints ]
        1     1    POSRES_FC_BB    POSRES_FC_BB    POSRES_FC_BB   
        5     1    POSRES_FC_BB    POSRES_FC_BB    POSRES_FC_BB   
        8     1    POSRES_FC_BB    POSRES_FC_BB    POSRES_FC_BB   
    ...
    #endif
    ```

* В файлах параметров каждого из шагов уравновешивания (`.mdp`) в строчке `define=` указано, какие ограничения будут включены на данном шаге. Например, в файле `step6.0_minimization.mdp`: 

    `define                  = -DPOSRES -DPOSRES_FC_BB=4000.0 -DPOSRES_FC_SC=2000.0 -DPOSRES_FC_LIPID=1000.0 -DDIHRES -DDIHRES_FC=1000.0
    `

**Схема уравновешиания** 

* На каждом цикле величина коэффициентов сил позиционных ограничений постепенно снижается:
* Первый цикл: шаг интегрирования 1 фс, чтобы снизить риск ошибок интегратора на неравновесной системе. Далее dt становится 2 фс.

### 2.4.2. Распределение усредненных по МД-траектории параметров системы

Для анализа систем "белок в мембране" используют следующие параметры:

* Профили плотности различных компонентов системы (белка, липидов, воды и т.д.)
  
* Параметры бислоя: 
  * Площадь на липидную головку
  * Порядок ацильных цепей по дейтерию
  * Толщина бислоя
  * Латеральная диффузия липидов
  
* Параметры белка: 
  * RMSD
  * вторичная структура пептида
  * наклон спирали
  
* Контакты белок-белок/белок-липиды/липид-липид: водородные связи, ионные мостики и т.д.
* Распределения гидрофобных и электростатических свойств на молекулярных поверхностях системы

И, конечно, многое другое :)

````python
from MDAnalysis.analysis import lineardensity

import numpy as np
import matplotlib.pyplot as plt
````

#### Информация о системе

````python
u = mda.Universe("step7_1.tpr", "step7_1_out.xtc")   # adapt to your files

print("Universe info:")
print("\nBox dimensions[x,y,z]: {}." \
      "".format(u.dimensions[:3]))

print("\n{} residues and {} atoms." \
      "".format(u.residues.n_residues, u.residues.n_atoms))

print("\nThe trajectory includes {} frames."\
      "".format(u.trajectory.n_frames))

protein  = u.select_atoms("protein")
print("\nProtein has {} residues." \
      "".format(protein.n_residues))

lipids  = u.select_atoms("not protein")
print("\nLipids molecules include {} residues and {} atoms." \
      "".format(lipids.n_residues, lipids.n_atoms))
````

#### Распределение средней плотности атомов вдоль оси Z

````python
lipids = u.select_atoms("resname POPC")      # or your lipid resnames
water  = u.select_atoms("resname TIP3")
headgroups    = lipids.select_atoms("name P N C12 C13 C14 C15")   # example
glycerol_ester = lipids.select_atoms("name C21 C22 C31 C32 O21 O31")  # example
acyl_chains   = lipids.select_atoms("name C2? C3? C4? C5? C6? C1?")   # all tail carbons

protein = u.select_atoms("protein")   # or a more specific selection


binsize = 1  # Å resolution along z

ld_head = lineardensity.LinearDensity(headgroups,
                                      grouping='atoms',
                                      binsize=binsize).run()
ld_gly  = lineardensity.LinearDensity(glycerol_ester,
                                      grouping='atoms',
                                      binsize=binsize).run()
ld_acyl = lineardensity.LinearDensity(acyl_chains,
                                      grouping='atoms',
                                      binsize=binsize).run()
ld_wat  = lineardensity.LinearDensity(water,
                                      grouping='atoms',
                                      binsize=binsize).run()

ld_prot = lineardensity.LinearDensity(
    protein,
    grouping='atoms',        # per-atom mass density
    binsize=binsize          # same binsize as for lipids/water
).run()


z_edges = ld_head.results.z['hist_bin_edges']          # shape (nbins+1,)
z_centers_A = 0.5 * (z_edges[1:] + z_edges[:-1])

# convert to nm
z = z_centers_A / 10.0

rho_head = ld_head.results.z['mass_density'] * 1000.0
rho_gly  = ld_gly.results.z['mass_density'] * 1000.0
rho_acyl = ld_acyl.results.z['mass_density'] * 1000.0
rho_wat  = ld_wat.results.z['mass_density'] * 1000.0
rho_prot = ld_prot.results.z['mass_density'] * 1000.0  # g/cm^3 → kg/m^3
````

````python
plt.figure(figsize=(6,4))

plt.plot(z, rho_head, 'k-', label="Headgroups")
plt.plot(z, rho_gly,  'r-', label="Glycerol Ester")
plt.plot(z, rho_acyl, 'g-', label="Acyl Chains")
plt.plot(z, rho_wat,  'b-', label="Water")
plt.plot(z, rho_prot, 'm-', label="Protein")   # magenta example

plt.xlabel("z-Coordinate (nm)")
plt.ylabel(r"Density (kg m$^{-3}$)")
plt.legend()
plt.tight_layout()
plt.show()
````

#### RMSD белка

````python
np.unique(protein.segids)
````

````python
def rmsf_from_calphas(ca):
    means = np.zeros((len(ca), 3))
    sumsq = np.zeros_like(means)

    for k, ts in enumerate(u.trajectory):
        sumsq += (k / (k + 1.0)) * (ca.positions - means)**2
        means[:] = (k * means + ca.positions) / (k + 1.0)

    return np.sqrt(sumsq.sum(axis=1) / (k + 1.0))

ca0 = u.select_atoms("segid seg_0* and protein and name CA")
ca1 = u.select_atoms("segid seg_1* and protein and name CA")

rmsf0 = rmsf_from_calphas(ca0)
rmsf1 = rmsf_from_calphas(ca1)

plt.figure(figsize=(6,4))
plt.plot(ca0.residues.resids, rmsf0, '.-', label="seg_0")
plt.plot(ca1.residues.resids - np.min(ca1.residues.resids), rmsf1, '.-', label="seg_1")

plt.ylim(0, 3.5)

plt.xlabel("residue number")
plt.ylabel(r"C$_\alpha$ RMSF (Å)")
plt.legend()
plt.tight_layout()
plt.show()
````
